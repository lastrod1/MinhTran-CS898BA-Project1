```python
import matplotlib.pyplot as plt
import numpy as np

data = np.load("../../data/font_digits.npz")
images, labels, font_ids = data["images"], data["labels"], data["font_ids"]

fig, axes = plt.subplots(4, 8, figsize=(12, 6))
idx = np.random.default_rng(1).choice(len(images), 32, replace=False)
for ax, i in zip(axes.flat, idx):
    ax.imshow(images[i], cmap="gray")
    ax.set_title(str(labels[i]))
    ax.axis("off")
plt.tight_layout()
plt.savefig("check.png")
```

```python
import sys
import os
import cv2 as cv
from src.grid_detection.sudoku_grid import solveSudoku, print_grid, draw_sudoku_grid
import matplotlib.pyplot as plt

grid_folder = os.path.abspath(os.path.join(os.getcwd(), "src", "grid_detection"))
sys.path.append(grid_folder)

from src.grid_detection.pipeline import pipeline
```

```python
IMAGE = "./images/raw_images/raw_4.png"

grid = pipeline(image_path=IMAGE)
grid_img = draw_sudoku_grid(grid)
```

```python
solved = solveSudoku(grid)

plt.subplot(1, 3, 1)
plt_img = plt.imread(IMAGE)
plt.imshow(plt_img)
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(grid_img)
plt.axis('off')


if solved == None:
    print("Grid not solvable")
else:
    print_grid(solved)
    
    plt.subplot(1, 3, 3)
    solved_img = draw_sudoku_grid(solved)
    plt.imshow(solved_img)
    plt.axis('off')

plt.show()
```

```python
def show_img(img, s, title): # s is for subplot
    plt.subplot(s[0], s[1], s[2])
    plt_img = cv.imread(img)
    plt.imshow(plt_img)
    plt.axis('off')
    plt.title(title)
```

```python
LINES_NB_IMG = "./images/test_images/lines_image_nb.png"
LINES_IMG = "./images/test_images/lines_image.png"
POINTS_IMG = "./images/test_images/points_image.png"
PREPROCESSED_IMG = "./images/test_images/preprocessed.png"
WARPED_IMG = "./images/test_images/warped.png"

show_img(IMAGE, (2,3,1), "Original")
show_img(WARPED_IMG, (2,3,2), "Warped")
show_img(PREPROCESSED_IMG, (2,3,3), "Preprocessed")
show_img(LINES_IMG, (2,3,4), "Lines")
show_img(LINES_NB_IMG, (2,3,5), "Lines (Not Binary)")
show_img(POINTS_IMG, (2,3,6), "Points")
plt.show()
```
